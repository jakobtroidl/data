```python
import numpy as np
import zarr
from ome_zarr.writer import write_image
```

```python
data = np.load("test.npy")  # Load your image data

print(data.shape)

# Create a Zarr store and root group
store = zarr.DirectoryStore("output.ome.zarr")
root = zarr.group(store=store)

# Write to the group with correct axes
write_image(
    image=data,
    group=root,
    axes="zyx",
    chunk_dims=(10, 340, 667),  # Optional
    dimension_separator="/"
)
```
